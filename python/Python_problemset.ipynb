*Write your explanation here.*

````python
# Example of the walrus operator
# Walrus operator(:=) allow you to assign a value to a variable and evaluate this value in the same expression.
# It make the code more concise and readable by eliminating the redundant function call.
def value(n):
    return n + 2
list = [ (a, b) for a in range(0, 11) if (b := value(a)) > 10]
print(list)
````

### 4c

Write a function called `my_zip(L1, L2, ..., Ln)` that takes $n \geq 2$ lists (all the same size) as input and returns a list of tuples just like `zip`. If $n$ is less than 2, then raise a `ValueError`.

````python
def my_zip(*lists):
    if len(lists) < 2:
        raise ValueError
    
    return [tuple(list[i] for list in lists) for i in range(len(list) + 1)]
    
# Example usage:
list1 = [1, 2, 3]
list2 = [4, 5, 6]
list3 = [1, 1, 1]

print(my_zip(list1, list2, list3))
````

### 4d

Using `my_zip`, write a function called `add_lists(L1, L2, ..., Ln)` that adds lists `L1` to `Ln` element-wise. You can assume they are all the same size (and non-empty), and all lists of numbers.

For example, `add_lists([1, 2, 3], [4, 5, 6], [1, 1, 1])` should return `[6, 8, 10]`.

````python
def add_lists(*lists):
    return [sum(item) for item in my_zip(*lists)]

print(add_lists(list1, list2, list3))  # [6, 8, 10]
````
